## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("=== Quick Look at the Starter Dataset ===\n")

# 1. Scale of the dataset
print(f"Total content pages: {len(df):,}")
print(f"Unique clients: {df['client_id'].nunique()}")

# 2. Pages with declining or flat trend
declining = df[df["trend_direction"].isin(["down", "flat"])]
print(f"Pages with declining or flat trend: {len(declining):,} ({len(declining)/len(df)*100:.1f}%)")

# 3. Pages with CTR opportunity
opportunity = df[
    (df["avg_position"] >= 5) &
    (df["avg_position"] <= 30) &
    (df["ctr"] < df["ctr"].median())
]

print(f"Pages ranking between positions 5–30 with below-median CTR: {len(opportunity):,}")
```

### Quick Look at the Data

The starter dataset contains **30,000** content pages from **32** clients, showing that manual review would be difficult at scale.

Approximately **58%** of pages have a declining or flat trend, indicating that many pages may require attention.

The dataset also contains **9,750** pages ranking between positions **5 and 30** with below-median CTR. These pages represent potential opportunities where improving content or search presentation could increase performance.

These observations support choosing the **Refresh / Content Opportunity Scoring** lane because the dataset contains enough scale and measurable opportunities for building a ranked decision-support system.
